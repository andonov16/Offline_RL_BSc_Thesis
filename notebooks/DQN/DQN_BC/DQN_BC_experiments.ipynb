```python
import os
import sys
import yaml
import pandas as pd
from optuna.terminator.improvement.emmr import torch

if 'google.colab' in sys.modules:
  from google.colab import drive
  drive.mount( "/content/drive")
  if os.path.isdir('drive/MyDrive/Offline_RL_BSc_Thesis_repo/Offline_RL_BSc_Thesis/notebooks/BC'):
    os.chdir('drive/MyDrive/Offline_RL_BSc_Thesis_repo/Offline_RL_BSc_Thesis/notebooks/BC')


project_root = os.path.abspath(os.path.join(os.path.dirname("__file__"), "../../../"))
if project_root not in sys.path:
    sys.path.append(project_root)

with open('../../../config/dqn_bc_experiments_config.yaml', 'r') as f:
    dqn_bc_experiments_config = yaml.safe_load(f)

torch.manual_seed(dqn_bc_experiments_config['experiment']['seed'])

selected_features = ['X', 'Y', 'lv_X', 'lv_Y', 'angle', 'angular_velocity', 'leg_1', 'leg_2', 'reward', 'next_X', 'next_Y', 'next_lv_X', 'next_lv_Y', 'next_angle', 'next_angular_velocity', 'next_leg_1', 'next_leg_2']
```

## Hyperparameter Tuning and Training

```python
from src.experiments.DQN_BC_experiments import conduct_dqn_bc_experiment
```

### Replay Buffer

```python
rb_train_df = pd.read_parquet('../../../data/replay_buffer_episodes/rb_train_next_states_included.parquet').drop(columns=['episode'])
rb_train_state_reward_nex_state_df = rb_train_df.drop(columns=['done'])
rb_train_dones_tensor = torch.tensor(rb_train_df['done']).float()

rb_normalization_techniques = {
    'raw': None,
    'max_abs': torch.jit.load(f'../../../models/BC/replay_buffer/normalization/max_abs_normalization.pt'),
    'min_max': torch.jit.load(f'../../../models/BC/replay_buffer/normalization/min_max_normalization.pt'),
    'robust': torch.jit.load(f'../../../models/BC/replay_buffer/normalization/robust_normalization.pt'),
    'standard': torch.jit.load(f'../../../models/BC/replay_buffer/normalization/standard_normalization.pt'),
}

rb_generative_models_scripts = {
    'raw': torch.jit.load(f'../../../models/BC/replay_buffer/BC_raw.pt'),
    'max_abs': torch.jit.load(f'../../../models/BC/replay_buffer/BC_max_abs.pt'),
    'min_max': torch.jit.load(f'../../../models/BC/replay_buffer/BC_min_max.pt'),
    'robust': torch.jit.load(f'../../../models/BC/replay_buffer/BC_robust.pt'),
    'standard': torch.jit.load(f'../../../models/BC/replay_buffer/BC_standard.pt'),
}
```

#### Raw Data

```python
norm_name = 'raw'

conduct_dqn_bc_experiment(
    dataset_name='replay_buffer',
    norm_technique_name=norm_name,
    norm_technique_script=rb_normalization_techniques[norm_name],
    generative_model_script=rb_generative_models_scripts[norm_name],
    selected_features=selected_features,
    output_model_name=f'DQN_BC_{norm_name.lower()}',
    train_df=rb_train_state_reward_nex_state_df,
    dones=rb_train_dones_tensor,
    experiments_config=dqn_bc_experiments_config,
)
```
